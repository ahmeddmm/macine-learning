```python
import numpy as np
import pandas as pd
import matplotlib.pylab as plt
import seaborn as sns
from sklearn.datasets import load_iris,load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score
```

```python
iris=load_iris()
iris
```

```python
df=pd.DataFrame(iris.data,columns=iris.feature_names)
df['target']=iris.target
```

```python
df.head()
```

```python
df.shape
```

```python
df.drop_duplicates(inplace=True)
df.info()
```

```python
df['target'].value_counts()
```

```python
corr=df.drop('target',axis=1).corr()
sns.heatmap(corr,annot=True,cmap='coolwarm')
```

```python

x=df.drop('target',axis=1)
y=df['target']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=42)
```

```python
tree=DecisionTreeClassifier(min_samples_split=40,max_depth=2,random_state=42)
tree.fit(x_train,y_train) 
```

```python
y_pred=tree.predict(x_test)
```

```python
acc=accuracy_score(y_test,y_pred)
print(f" acc: {acc:.2f}")
```

```python
print(f"depth : {tree.get_depth()}")
```

```python
print(f"leaves:{tree.get_n_leaves()}")
```

```python
plt.figure(figsize=(12,8))
plot_tree(tree,feature_names=iris.feature_names,class_names=iris.target_names,rounded=True,filled=True,fontsize=8)
plt.show()
```
