```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_diabetes
from sklearn.metrics import accuracy_score,mean_absolute_error,mean_squared_error
```

```python
#df=pd.DataFrame(diabites.data,columns=diabites.feature_names)
#df['target']=diabites.target
df=pd.read_csv('diabetes.csv')
```

```python
df.shape
df.head()
```

```python
corr=df.corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr,cmap='coolwarm',annot=True)
```

```python
x1=df.drop('Outcome',axis=1)
y1=df['Outcome']
```

```python
x_trained,x_test,y_train,y_test=train_test_split(x1,y1,test_size=.2,random_state=42)
```

```python

scaller=StandardScaler()
x_trained_scalled=scaller.fit_transform(x_trained)
x_test_scalled=scaller.transform(x_test)
```

```python

#model=LinearRegression()
model=LogisticRegression()
model.fit(x_trained_scalled,y_train)
```

```python
y_pred=model.predict(x_test_scalled)
y_pred
```

```python
# mae=mean_absolute_error(y_test,y_pred)
# mae
acc=accuracy_score(y_test,y_pred)
acc
```
