```python
arr=np.random.randint(0,99,(4,5))
print("the matrix",arr)
print("shape",arr.shape)
print("size",arr.size)
print("number of dimentions",arr.ndim)
print("firist row",arr[0,:])
print("last column",arr[:,4])
print("values greater than 50",arr[arr>50])
print("values less than 30",np.where(arr<30,0,arr))
print("to one dimentional array",arr.reshape(1,20))
print("original shape",arr.reshape(4,5))
```

```python
arr=np.array(range(10,121,10))
print("array\n",arr)
print("shape\n",arr.shape)
print("size\n",arr.size)
print("n of dimentions\n",arr.ndim)
print("data type",arr.dtype)
print("greater than 60\n",arr[arr>60])
print("replace greater than 90 with 999",np.where(arr>90,999,arr))
print(arr.flatten())
arr=arr.flatten()
print(arr[[1,2,3]])
```
