```python
class product:
    def __init__(self,name,price,quantity):
        self.name=name
        self.price=price
        self.quantity=quantity
    def restock(self,q):
        self.quantity=self.quantity+q
    def sell(self,q):
        if q > self.quantity:
            print("not enough stouck")
        else:
            self.quantity=self.quantity+q
            print(f"price is {q*self.price}")
    
```

```python
computer=product("computer",30,100)
mouse=product("mouse",12,99)
mouse.sell(2)
mouse.sell(2100)
```

```python
class student:
    def __init__(self,name,id):
        self.name=name
        self.id=id
        self.couse=[]
    def add_course(self,course):
        self.couse.append(course)
    def display(self):
        print(f"student name :{self.name}")
        print("courses:")
        for _ in self.couse:
            print("   "+_)

        
```
